```python
import pandas as pd
import numpy as np
import yaml
import os
```

```python
os.chdir('..')
```

```python
config_path = os.path.join('config', 'config.yaml')

with open(config_path, 'r') as file:
    config = yaml.safe_load(file)
config
```

```python
raw_file = os.path.join(config['DATA_DIR']['DATA_100K']['RAW'], 'ratings.csv')


ratings_df = pd.read_csv(raw_file)
ratings_df.head()
```

```python
# dataset shape
ratings_df.shape
```

```python
n_movies = ratings_df['movieId'].nunique()
n_users = ratings_df['userId'].nunique()


print(f'Number of movies in the dataset: {n_movies}')
print(f"Number of users in the dataset: {n_users}")
```

```python
summary_movies_watched = (
    ratings_df
        [['userId', 'movieId']]
        .groupby('userId')
        .agg('count')
        .describe()
)
summary_movies_watched.columns = ['movies_watched']
summary_movies_watched
```
